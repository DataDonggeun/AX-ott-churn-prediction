# 컬럼명 변경

```python
import pandas as pd

membership = pd.read_csv(
    r"C:\myCode\ott-churn-prediction\kim.kwangil\data\01_raw\Membership_train.csv"
).rename(
    columns={
        "uno": "USER_KEY",
        "productcode": "product_code",
        "pgamount": "price",
        "chargetypeid": "billing_method",
        "concurrentwatchcount": "max_screen",
        "promo_100": "is_promotion",
        "coinReceived": "is_churn_prevented",
        "devicetypeid": "payment_device",
        "isauth": "is_user_verified",
        "gender": "gender",
        "agegroup": "age",
        "registerday": "reg_date",
        "registerhour": "reg_hour",
        "endday": "end_date",
        "Repurchase": "is_repurchase",
    }
)

user_mapping = pd.read_csv(
    r"C:\myCode\ott-churn-prediction\kim.kwangil\data\01_raw\mapping.csv"
).rename(
    columns={
        "uid": "USER_KEY",
        "USER_ID": "USER_NUM",
    }
)

view_history = pd.read_csv(
    r"C:\myCode\ott-churn-prediction\kim.kwangil\data\01_raw\Views_train.csv"
).rename(
    columns={
        "USER_ID": "USER_NUM",
        "MOVIE_ID": "MOVIE_NUM",
        "DURATION": "watch_time(min)",
        "WATCH_DAY": "watch_day",
        "WATCH_SEQ": "watch_seq",
    }
)

movie_master = pd.read_csv(
    r"C:\myCode\ott-churn-prediction\kim.kwangil\data\01_raw\Movies.csv"
).rename(
    columns={
        "MOVIE_ID": "MOVIE_NUM",
        "TITLE": "movie_title",
        "RELEASE_MONTH": "ott_release_month",
        "Category": "category",
    }
)
```

# 저장

```python
from pathlib import Path

output_dir = Path(r"C:\myCode\ott-churn-prediction\kim.kwangil\data\02_interim")
output_dir.mkdir(parents=True, exist_ok=True)

membership.to_csv(output_dir / "membership.csv", index=False)
user_mapping.to_csv(output_dir / "user_mapping.csv", index=False)
view_history.to_csv(output_dir / "view_history.csv", index=False)
movie_master.to_csv(output_dir / "movie_master.csv", index=False)
```
