# 왓챠피디아 평점 크롤링

Movie_Master의 영화 제목으로 왓챠피디아를 검색해 평점을 수집합니다.

- 제목 유사도 + 연도 기반 매칭
- `ratings_avg`, `ratings_count` 수집
- 출력: `data/02_interim/260506_watcha_ratings/watcha_ratings.csv`

```python
import random
import re
import time
import requests
import pandas as pd
from pathlib import Path
from difflib import SequenceMatcher

print('라이브러리 로드 완료')
```

```python
# ★ _c_pdi 쿠키 값 만료되면 여기만 갱신
DEVICE_ID = 'web-vPnHq8ZT76Fuz7NZwVJIQrytV8YU6D'

HEADERS = {
    'accept'                      : 'application/vnd.frograms+json;version=2.1.0',
    'accept-language'             : 'ko-KR,ko;q=0.9',
    'user-agent'                  : 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36',
    'x-frograms-app-code'         : 'Galaxy',
    'x-frograms-client'           : 'Galaxy-Web-App',
    'x-frograms-client-version'   : '2.1.0',
    'x-frograms-device-identifier': DEVICE_ID,
    'x-frograms-galaxy-language'  : 'ko',
    'x-frograms-version'          : '2.1.0',
    'referer'                     : 'https://pedia.watcha.com',
}

BASE     = Path('..').resolve()
DATA_IN  = BASE / 'data/02_interim/260504_promotion_split/promotion_0_movie_master_v2.csv'
DATA_OUT = BASE / 'data/02_interim/260506_watcha_ratings'
DATA_OUT.mkdir(exist_ok=True)

SLEEP_SEC  = 0.4
SIM_THRESH = 0.5

print('설정 완료')
print('입력 파일:', DATA_IN)
print('출력 폴더:', DATA_OUT)
```

```python
def clean_title(t: str) -> str:
    t = re.sub(r'[\-\(\)\[\]【】（）：:]', ' ', str(t))
    return re.sub(r'\s+', ' ', t).strip()

def remove_spaces(t: str) -> str:
    return t.replace(' ', '')

def extract_year(t: str):
    m = re.search(r'\((\d{4})\)', str(t))
    return int(m.group(1)) if m else None

def similarity(a: str, b: str) -> float:
    return SequenceMatcher(None, a.lower(), b.lower()).ratio()

def best_match(candidates, title, year):
    best, best_score = None, 0.0
    for c in candidates:
        if c.get("content_type") != "movies":
            continue
        sim1 = similarity(clean_title(title), clean_title(c.get("title", "")))
        sim2 = similarity(remove_spaces(title), remove_spaces(c.get("title", "")))
        sim  = max(sim1, sim2)
        yr_penalty = -abs(year - c["year"]) * 0.05 if (year and c.get("year")) else 0
        score = sim + yr_penalty
        if score > best_score:
            best_score, best = score, c
    return best if best_score >= SIM_THRESH else None

def search_watcha(title: str, year=None):
    queries = list(dict.fromkeys([title, remove_spaces(title)]))
    candidates = []
    try:
        for q in queries:
            r = requests.get(
                "https://pedia.watcha.com/api/searches",
                params={"query": q, "type": "contents"},
                headers=HEADERS, timeout=10,
            )
            if r.status_code != 200:
                continue
            result = r.json().get("result", {})
            for key in ["top_results", "contents", "movies"]:
                candidates += result.get(key, [])
            time.sleep(random.uniform(1, 2))
        return best_match(candidates, title, year)
    except Exception as e:
        print(f"  [검색 오류] {e}")
        return None

def get_rating(code: str):
    try:
        r = requests.get(
            f"https://pedia.watcha.com/api/contents/{code}",
            headers=HEADERS, timeout=10,
        )
        if r.status_code != 200:
            print(f"  [HTTP오류] {r.status_code}")
            return None, None
        result = r.json().get("result", {})
        avg = result.get("ratings_avg")
        cnt = result.get("ratings_count")
        avg = round(avg / 2, 4) if avg is not None else None  # 10점 -> 5점
        return avg, cnt
    except Exception as e:
        print(f"  [평점 오류] {e}")
        return None, None

print("함수 정의 완료")
```

## 테스트 (5개)

```python
p0 = pd.read_csv(BASE / 'data/02_interim/260504_promotion_split/promotion_0_movie_master_v2.csv', encoding='utf-8-sig')
p1 = pd.read_csv(BASE / 'data/02_interim/260504_promotion_split/promotion_1_movie_master_v2.csv', encoding='utf-8-sig')

all_df = pd.concat([p0, p1]).drop_duplicates(subset='MOVIE_NUM').reset_index(drop=True)

# ★ 여기만 바꾸세요: 1 / 2 / 3
WORKER_ID    = 1
TOTAL_WORKERS = 3

chunk = len(all_df) // TOTAL_WORKERS
start = (WORKER_ID - 1) * chunk
end   = len(all_df) if WORKER_ID == TOTAL_WORKERS else start + chunk

df = all_df.iloc[start:end].reset_index(drop=True)
print(f'전체: {len(all_df):,}개  |  워커 {WORKER_ID}/{TOTAL_WORKERS}: {start}~{end-1} ({len(df):,}개)')
df.head()
```

```python
# 5개 테스트
for _, row in df.head(5).iterrows():
    raw   = row["movie_title"]
    year  = extract_year(raw)
    title = clean_title(raw)
    match = search_watcha(title, year)
    if match:
        avg, cnt = get_rating(match["code"])
        if avg:
            print(f"✅ {raw} → {match['title']} ({match['year']}) | 평점: {avg:.2f}/5 ({cnt:,}명)")
        else:
            print(f"⚠️  {raw} → {match['title']} | 평점 없음")
    else:
        print(f"❌ {raw} → 매칭 실패")
    time.sleep(0.5)
```

## 전체 크롤링 (약 3시간)

```python
SLEEP_MIN, SLEEP_MAX = 4, 6
```

```python
results = []

for i, row in df.iterrows():
    raw   = row['movie_title']
    year  = extract_year(raw)
    title = clean_title(raw)
    match = search_watcha(title, year)

    if match:
        avg, cnt = get_rating(match['code'])
        results.append({
            'MOVIE_NUM'    : row['MOVIE_NUM'],
            '원제'         : raw,
            '왓챠_제목'    : match.get('title'),
            '왓챠_연도'    : match.get('year'),
            'watcha_code'  : match.get('code'),
            'ratings_avg'  : avg,
            'ratings_count': cnt,
        })
    else:
        results.append({
            'MOVIE_NUM'    : row['MOVIE_NUM'],
            '원제'         : raw,
            '왓챠_제목'    : None,
            '왓챠_연도'    : None,
            'watcha_code'  : None,
            'ratings_avg'  : None,
            'ratings_count': None,
        })

    if (i + 1) % 100 == 0:
        matched = sum(1 for r in results if r['ratings_avg'] is not None)
        print(f'[{i+1:,}/{len(df):,}] 매칭: {matched}개 ({matched/(i+1)*100:.1f}%)')

    time.sleep(random.uniform(SLEEP_MIN, SLEEP_MAX))

print('크롤링 완료!')
```

```python
result_df = pd.DataFrame(results)
out_path  = DATA_OUT / f'watcha_ratings_worker{WORKER_ID}.csv'
result_df.to_csv(out_path, index=False, encoding='utf-8-sig')

matched = result_df['ratings_avg'].notna().sum()
print(f'매칭 성공: {matched:,}/{len(result_df):,}개 ({matched/len(result_df)*100:.1f}%)')
print(f'저장 완료: {out_path}')
result_df.head(10)
```
